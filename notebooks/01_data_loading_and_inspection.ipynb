# Data Loading and Initial Inspection

## Objective

This notebook loads the raw Orders, Returns, and People datasets and verifies that the source files are ready for analysis.

The notebook covers:

- Project and raw-data path setup
- Source-file availability checks
- CSV data loading with Pandas
- Dataset dimensions
- Initial record previews

No data cleaning or transformation is performed in this notebook.

## 1. Environment and Path Setup

I import the required libraries and define reusable project-relative paths.

```python
from pathlib import Path
import pandas as pd

current_folder = Path.cwd()
project_root = current_folder.parent
raw_data_folder = project_root / "data" / "raw"

print("Notebook folder:", current_folder.name)
print("Project folder:", project_root.name)
print(
    "Raw data folder:",
    raw_data_folder.relative_to(project_root)
)
print("Raw folder exists:", raw_data_folder.exists())
```

## 2. Source File Validation

I verify that the expected raw CSV files are available before loading the datasets.

```python
raw_files = sorted(raw_data_folder.glob("*.csv"))

for file in raw_files:
    print(file.name)

print("Total CSV files:", len(raw_files))
```

```python
orders_path = raw_data_folder / "orders_data.csv"
returns_path = raw_data_folder / "returns_data.csv"
people_path = raw_data_folder / "people_data.csv"
```

## 3. Dataset Loading and Initial Preview

I load the Orders, Returns, and People datasets and validate their dimensions and initial records.

```python
orders = pd.read_csv(
    orders_path,
    dtype={"Postal Code": "string"}
)

print("Orders shape:", orders.shape)
display(orders.head())
```

```python
returns = pd.read_csv(
    returns_path
)

print("Returns shape:", returns.shape)
display(returns.head())
```

```python
people = pd.read_csv(
    people_path
)

print("People shape:", people.shape)
display(people.head())
```
